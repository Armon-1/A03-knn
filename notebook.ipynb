```python
## The difference between regression and classification is that regression is used to predict a number 
## and classification is used to try to predict a category or class. For example, predicting a 
## specific number like income would be regression while predicting something like whether or not it 
## is Monday is classification.
```

```python
## A confusion table is something that shows the user how many predictions the model got right and 
## wrong. It helps us understand which classes the model predicts well and which are making mistakes.
```

```python
## A SSE (which is a sum of squared errors) quantifies how far the model's prediction is from the 
## actual value. It squares all the errors and adds them together.
```

```python
## Overfitting is when a model fits the data with too much detail and too closely. 
## Underfitting is when a model is too simple and doesn't capture the important patterns in data.
```

```python
## Splitting the data into training and testing sets allows us to train the model on a set and then 
## next see how it performs on data that it hasn't seen yet. Trying different values of k and 
## comparing their accuracy or SSE helps us choose the k that gives the best results which improves 
## the strength of the model performance.
```

```python
## Some strengths of both are that a class label prediction gives one final category which is easy 
## to use and pretty simple for the user. A probability distribution gives probabilities for each 
## class and this is gives some more information about any uncertainties. Some weaknesses are that a 
## class label prediction doesn't show how confident a model is. A probability distribution is a bit 
## harder to read.
```

```python
import pandas as pd

mines = pd.read_csv("./data/land_mines.csv")
mines.head()
```

```python
mines.describe()
```

```python
mines["mine_type"].value_counts()
```

```python
## There are 338 observations and 5 mine types. The classes are fairly balanced, with about 
## 65-71 observations in each class. The features are voltage, height, and soil.
```

```python
from sklearn.model_selection import train_test_split

X = mines[["voltage", "height", "soil"]]
y = mines["mine_type"]

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.5, random_state=42
)
```

```python
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score

for k in range(1, 11):
    model = KNeighborsClassifier(n_neighbors=k)
    model.fit(X_train, y_train)
    predictions = model.predict(X_test)
    print(k, accuracy_score(y_test, predictions))
```

```python
## I tested the k values from 1 to 10. I selected k = 2 because it had the highest test accuracy 
## of  40.2%.
```

```python
model = KNeighborsClassifier(n_neighbors=2)
model.fit(X_train, y_train)

predictions = model.predict(X_test)

pd.crosstab(y_test, predictions, rownames=["Actual"], colnames=["Predicted"])
```

```python
accuracy_score(y_test, predictions)
```

```python
## It appears that the model is around 40.2% accurate overall. It performs best on the mine type 2, 
## where 25 out of 33 were predicted correctly. It performs the most poorly on mine type 5, 
## where only 3 out of 31 were predicted correctly.
```

```python
## Since the model makes a lot of mistakes on some mine types, this makes me wary about using it by 
## itself for any final decision. I'd probably use it as a first prediction tool but also have a 
## person standing by to verify it. This is especially true in the case of mine 4 and 5 where it is 
## not very accurate at all.
```

```python
## Q1.2

## A confusion table breaks down predictions by actual class and predicted class. 
## Correct predictions appear along the diagonal, while the other cells show the specific 
## types of mistakes the classifier is making.
```

```python
## Q1.3

## SSE measures the total squared prediction error in a model. Each prediction error is squared and 
## then added together, so larger errors are penalized more heavily. Smaller SSE means the 
## predicted numerical values are closer overall to the true values.
```

```python
## Q1.4

## Overfitting means the model learns the training data so specifically that it may also learn 
## noise, which hurts performance on new observations. Underfitting is the opposite problem: 
## the model is not flexible enough to capture the important structure in the data.
```

```python
## Q1.5

## The training set is used to fit the model, while separate data are needed to judge how well the 
## model generalizes. If different values of k are compared using the same test set, that set is 
## effectively being used for validation, so a separate final test set would give a cleaner estimate 
## of performance on unseen data.
```

```python
## Q1.6

## A class-label prediction tells us only which class the model chooses. That makes the result 
## easy to act on, but it hides how uncertain the model may be. A probability output shows how the 
## model divides its confidence across the possible classes, which is more informative but requires 
## more interpretation.
```

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score
```

```python
X = mines[["voltage", "height", "soil"]]
y = mines["mine_type"]

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.5, random_state=42, stratify=y
)
```

```python
X_fit, X_val, y_fit, y_val = train_test_split(
    X_train, y_train, test_size=0.25, random_state=42, stratify=y_train
)

scaler = StandardScaler()

X_fit_scaled = scaler.fit_transform(X_fit)
X_val_scaled = scaler.transform(X_val)
```

```python
for k in range(1, 21):
    model = KNeighborsClassifier(n_neighbors=k)
    model.fit(X_fit_scaled, y_fit)
    predictions = model.predict(X_val_scaled)
    print(k, accuracy_score(y_val, predictions))
```

```python
## k = 3 and k = 9 had the highest validation accuracy, so I chose k = 3 as the simpler option.
```

```python
scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

model = KNeighborsClassifier(n_neighbors=3)
model.fit(X_train_scaled, y_train)

predictions = model.predict(X_test_scaled)
```

```python
print("Accuracy:", accuracy_score(y_test, predictions))

pd.crosstab(
    y_test,
    predictions,
    rownames=["Actual"],
    colnames=["Predicted"]
)
```
